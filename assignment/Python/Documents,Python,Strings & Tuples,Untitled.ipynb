# String Slicing and Indexing

```python
s = "Hello Zara, welcome to Python programming"
```

```python
print("First character:", s[0])
```

```python
print("Last character:", s[-1])
```

```python
print("First 5 characters:", s[:5])
```

```python
print("Last 11 characters:", s[-11:])
```

```python
print("Reverse string:", s[::-1])
```

```python
start = s.find("Python")
```

```python
print("Word 'Python':", s[start:start+6])
```

# String Methods

```python
strM = "Python beginner tutorial"
```

```python
print("Uppercase:", strM.upper())
```

```python
print("Lowercase:", strM.lower())
```

```python
print("Capitalize:", strM.capitalize())
```

```python
print("Count of 't':", strM.count('t'))
```

```python
print("After Replace:", strM.replace("Python", "Machine Learning"))
```

# Tuples

```python
t1 = (10, 20, 30)
```

```python
t2 = (40, 50, 60)
```

```python
t_combine = t1 + t2
```

```python
print("Concatenated Tuple:", t_combine)
```

```python
print("Repeated Tuple:", t_combine * 3)
```

```python
print("3rd Element:", t_combine[2])
```

```python
print("First Three Elements:", t_combine[:3])
```

```python
print("Last Three Elements:", t_combine[-3:])
```
